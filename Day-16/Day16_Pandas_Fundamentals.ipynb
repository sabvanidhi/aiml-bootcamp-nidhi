```python
#confirm age column it's have any null values or not
df['age'].isna().sum()
```

```python
#Strategy 1 — drop: remove rows with missing age. How many rows remain? What did the mean age become?
df_drop = df.dropna(subset=["age"])

print("len of raw:", len(df_drop))
print("Mean age:", df_drop["age"].mean())
print("Standard deviation:", df_drop["age"].std())
```

```python
#Strategy 2 — mean fill: fill missing ages with the overall mean age.
#Check what happens to the standard deviation, and explain why it changes.
df_mean = df.copy()

df_mean["age"] = df_mean["age"].fillna(df_mean["age"].mean())

print("Mean age:", df_mean["age"].mean())
print("Standard deviation:", df_mean["age"].std())
```

```python
# Strategy 3 — group fill: fill each missing age with the median age of passengers in the same class and sex.
df_group = df.copy()

df_group["age"] = df_group.groupby(
    ["pclass", "sex"]
)["age"].transform(lambda x: x.fillna(x.median()))

print("Mean age:", df_group["age"].mean())
print("Standard deviation:", df_group["age"].std())
```

```python
comparison = pd.DataFrame({
    "Strategy": [
        "Drop missing ages",
        "Mean fill",
        "Class + Sex median fill"
    ],
    "Mean Age": [
        df_drop["age"].mean(),
        df_mean["age"].mean(),
        df_group["age"].mean()
    ],
    "Standard Deviation": [
        df_drop["age"].std(),
        df_mean["age"].std(),
        df_group["age"].std()
    ]
})

print(comparison)
```

# Task-3

The three missing-age strategies each have different advantages and disadvantages. Dropping missing ages is simple and avoids adding artificial values, but it removes about 20% of the passengers from the age analysis. Mean filling keeps all passengers but reduces the variation because every missing age receives the same value. Group median filling is more representative because it uses the passenger's class and sex when estimating the missing age. For this dataset, I would choose group median filling because it keeps all rows while using more relevant information than a single overall mean. There is no single correct strategy because the best choice depends on the goal; for example, dropping rows may be reasonable when only a small number of values are missing, while group-based imputation may be better when preserving the dataset is important. Clearly documenting an imputation method is important because hiding that a large portion of the data was changed could make the analysis misleading and reduce trust in the results.
