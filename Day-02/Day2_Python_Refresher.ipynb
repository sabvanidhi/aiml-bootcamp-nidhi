```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        self.marks=[]

    def add_mark(self,score):
        self.marks.append(score)

    def average(self):
        return sum(self.marks)/len(self.marks)

    def highest(self):
        return max(self.marks)

    def lowest(self):
        return min(self.marks)

student1=student('abc',20)
student2=student('xyz',23)
student3=student('pqr',19)


student1.add_mark(80)
student1.add_mark(60)
student1.add_mark(70)
student1.add_mark(89)

student2.add_mark(90)
student2.add_mark(69)
student2.add_mark(76)
student2.add_mark(88)

student3.add_mark(99)
student3.add_mark(86)
student3.add_mark(56)
student3.add_mark(77)

print(student1.name,student1.age,student1.marks,student1.highest(),student1.average(),student1.lowest())
print(student2.name,student2.age,student2.marks,student2.highest(),student2.average(),student2.lowest())
print(student3.name,student3.age,student3.marks,student3.highest(),student3.average(),student3.lowest())
    
```

```python
#__repr__:when we pass a list of student then we use 
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll
        self.marks = []

    def __str__(self):
        return f"name: {self.name}, roll: {self.roll}"

    def __repr__(self):
        return f"name:{self.name},roll:{self.roll}"

    def __len__(self):
        return len(self.marks)

    def __eq__(self):
        return self.roll==self.roll

s1 = Student("Asha", 12)
s2 = Student("Nidhi", 13)

print(s1)
print(s2)

students = [s1, s2]
print(students)
```

```python
def fib_list(n):
    result = []
    a = 0
    b = 1
    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result

print(fib_list(10))
```

```python
def fib_gen(n):
    a = 0
    b = 1

    for _ in range(n):
        yield a

        a, b = b, a + b

print("Generator:")
print(list(fib_gen(10)))
```

```python

gen = fib_gen(5)

print("First loop:")
for x in gen:
    print(x)

print("Second loop:")
for x in gen:
    print(x)
```

```python
def class_average(students, passing=40):
    total = 0
    passed = 0

    for s in students:


        total += s["marks"]

        if s["marks"] > passing:
            passed += 1

    average = total / len(students)
    pass_rate = passed / len(students) * 100

    return average, pass_rate


data = [
    {"name": "asha", "marks": 88},
    {"name": "ravi", "marks": 92},
    {"name": "meera", "marks": 79},
    {"name": "dev", "marks": 40},
]

avg, rate = class_average(data)

print(f"average: {avg}")
print(f"pass rate: {rate}%")
```
