```python
# Plot the distribution of SalePrice (histogram)
sns.histplot(df.SalePrice,kde=1)
plt.title('Distribution of SalePrice')
plt.xlabel("sales price")
plt.ylabel('num of house')

#SalePrice:The distribution is right-skewed. Most houses have lower-to-middle prices, while a smaller number of houses have very high prices.
```

# Task-3

```python
#Scatter plot: GrLivArea vs SalePrice.
sns.scatterplot(data=df, x= "GrLivArea",  y="SalePrice")

plt.title("Above-Ground Living Area vs Sale Price")
plt.xlabel("Above-Ground Living Area")
plt.ylabel("Sale Price")

#SalePrice generally increases as GrLivArea increases, showing a positive relationship between living area and house price, although some houses do not follow the overall pattern.
```

```python
# Boxplot: SalePrice across OverallQual levels.
sns.boxplot(data=df , x="OverallQual", y="SalePrice")
plt.title("boxplot of saleprice vs overallqual")
plt.xlabel('OverallQual')
plt.ylabel('SalePrice')

#SalePrice generally increases as OverallQual increases, indicating a strong positive relationship between overall house quality and price.
```

```python
#Boxplot: SalePrice by Neighborhood
sns.boxplot( data=df ,x="Neighborhood", y="SalePrice" )
plt.title('boxplot of SalePrice vs Neighborhood')
plt.xlabel('SalePrice')
plt.ylabel('Neighborhood')
plt.xticks(rotation=45, ha="right")

#SalePrice varies substantially across neighborhoods, and sorting by median price makes the differences in typical house prices between neighborhoods easier to compare.
```

```python
sns.scatterplot( data=df, x="YearBuilt", y="SalePrice" )
plt.title("scatter plot of the YearBuilt vs SalePrice")
plt.xlabel('YearBuilt')
plt.ylabel('SalePrice')
```

```python
#boxplot of SalePrice column and show a outlier
sns.boxplot(df.SalePrice)
plt.title('boxplot of SalePrice')
plt.xlabel('SalePrice')
```

```python
#calculate a IQR
q1=df['SalePrice'].quantile(0.25)
q3=df['SalePrice'].quantile(0.75)

q1,q3
```

```python
IQR=q3-q1
IQR
```

```python
lower_bound = q1 - 1.5 * IQR
upper_bound = q3 + 1.5 * IQR

lower_bound , upper_bound
```

```python
price_outlier=(df['SalePrice']<lower_bound) | (df['SalePrice'] > upper_bound)
len(price_outlier)
```

```python
df[(df['GrLivArea'] > 4000) & (df['SalePrice'] < 200000)]
```

```python
#A small number of houses may have very large living areas but relatively low sale prices, making them unusual compared with the general positive relationship between GrLivArea and SalePrice.
```

```python
from plotnine import ggplot, aes, geom_point, labs
```

```python
plot = (
    ggplot(
        df,
        aes(
            x="GrLivArea",
            y="SalePrice",
            color="OverallQual"
        )
    )
    + geom_point(alpha=0.4)
    + labs(
        title="Above-Ground Living Area vs Sale Price",
        x="Above-Ground Living Area (sq ft)",
        y="Sale Price",
        color="Overall Quality"
    )
)

plot
```
