```python
#create a list of squares using both loop and functional programming
abc=[10,20,30,40,50]
#loop
for n in abc:
    a=n*n
    print(a)
#functional
a=[n*n for n in abc]
print(a)
```

```python
#create a list of words and parform append and strip
a = ['python ', 'html', 'css', 'java']

# Using loops
b = []
c = []

for n in a:
    b.append(n.upper())

print(b)
for n in a:
    c.append(n.strip())

print(c)

# Using functional/comprehension style
abc = [n.upper() for n in a]
pqr = [n.strip() for n in a]

print(abc)
print(pqr)
```

```python
#multiplication of two list
a = [1, 2, 3, 4]
b = [5, 6, 7, 8]

# Loop
result = []

for x, y in zip(a, b):
    result.append(x * y)

print(result)


# Functional
z = list(map(lambda x, y: x * y, a, b))

print(z)
```

```python
#append a word which length is >4
b = ['python', 'css', 'html', 'java', 'mathametic']

# Loop
result = []

for word in b:
    if len(word) > 4:
        result.append(word)

print(result)


# Functional
a = list(filter(lambda word: len(word) > 4, b))

print(a)
```

```python
#used a reduce function to add all the numbers in a list
from functools import reduce
#loop
#functional
a=[10,20,30,40,50,60]

b=reduce(lambda x,y:x+y,a)
print(b)
```

```python
#sum of prices in a list of dictionaries
list=[
    {"item":"abc","price":30},
    {"item":"xyz","price":90},
    {"item":"aaa","price":76},
    {"item":"bbb","price":40}
    ]
#loop
#functional
b=sum(map(lambda list:list["price"],list))

print(b)
```

```python
#flatten a list of lists using both loop and functional programming
from itertools import chain

# Loop
a = [[1, 2], [3, 4], [5]]
result = []

for sublist in a:
    for item in sublist:
        result.append(item)

print(result)


# Functional
z = list(chain([1, 2], [3, 4], [5]))

print(z)
```

```python
from collections import Counter
#loop
#functional
b=['python','css','html','java','mathametic']

a=dict(map(lambda x:(x,len(x)),b))

print(a)
```

```python
#remove empty strings and None values from a list
l1 = ["", None, [], 'abc', 'def', 'xyz']

# Loop
result = []

for item in l1:
    if item:
        result.append(item)

print(result)


# Functional
z = list(filter(None, l1))

print(z)
```

```python
#sum of numbers in a list using both loop and functional programming
from itertools import accumulate

abc = [10, 20, 30, 40, 50, 60]

# Loop
result = []
total = 0

for num in abc:
    total += num
    result.append(total)

print(result)


# Functional
z = list(accumulate(abc))

print(z)
```

```python
# remove stopwords and count word frequency
import string

with open("abc.txt", "r") as f:
    text = f.read().lower()

    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )
    words = text.split()
    stopwords = {
    "the", "a", "an", "is", "and",
    "of", "to", "in", "for", "on"
}
new_words = []
for i in words:
    if i not in stopwords:
        new_words.append(i)

words = new_words
counter = Counter(words)
print(counter)
print(counter.most_common(3))
```

```python
#combinations, permutations, and product of a list
from itertools import combinations,permutations,product

student=["abc","xyz","pqr","lmn","def"]

print("Combinations of 2 students:")
print(list(combinations(student, 2)))

print("Combinations of 3 students:")
print(list(combinations(student, 3)))

print("Permutations of 2 students:")
print(list(permutations(student, 2)))

print("Product of colors and sizes:")
print(list(product([["red"], ["blue"], ["green"]],[["small"], ["medium"], ["large"]])))
```

```python
#email check
import re

text = """
Casha.sharma@codetrade.io wrote to ravi_k99@gmail.com
meera.p+work@company.co.in, phone 022-2555-1234
dev@sub.domain.example.org and +91 98765 43210
not.an.email@ nor @nothing.com -- watch these
"""

emails = re.findall(r"[\w.+-]+@[\w-]+\.[\w.]+", text)
print("emails found:")
for e in emails:
    print("  ", e)


phones = re.findall(
    r'(?:\+91[\s-]?\d{5}[\s-]?\d{5}|\b0\d{2,4}[-\s]\d{3,4}[-\s]\d{4}\b)',
    text
)
print(phones)


parts = re.findall(r'([\w.+-]+)@([\w.-]+\.\w+)', text)
print("\nUsername and domain:")
print(parts)

domains = [domain for username, domain in parts]
domain_counts = Counter(domains)
print("\nDomain counts:")
print(domain_counts)
```

```python
#create a list and acc 
from collections import defaultdict

employees = [
    ("IT", "abc"),
    ("HR", "def"),
    ("IT", "xyz"),
    ("Sales", "pqr"),
    ("HR", "aaa")
]

departments = defaultdict(list)

for department, employee in employees:
    departments[department].append(employee)

print(departments)
```

```python
#namedtuple example
from collections import namedtuple

Point = namedtuple("Point", ["name", "age", "city"])

people = [
    Point("abc", 20, "Ahmedabad"),
    Point("xyz", 22, "Delhi"),
    Point("pqr", 29, "Mumbai"),
    Point("aba", 30, "Jaipur")
]

for p in people:
    print(p)
    print("p.name =", p.name, "p.age =", p.age, "p.city =", p.city)
    print("still a tuple:", p[0], p[1], p[2])
    print("unpacks like one:", *p)
    print()  
```

```python
#open a log file and write some log messages into it
from datetime import datetime, timedelta
import random

levels = ["INFO", "INFO", "INFO", "WARN", "ERROR"]

messages = {
    "INFO": [
        "User logged in",
        "Data processing started",
        "File uploaded",
        "Request completed",
        "User logged out"
    ],
    "WARN": [
        "Memory usage is high",
        "Slow response detected",
        "Disk space is getting low",
        "Too many requests"
    ],
    "ERROR": [
        "Database connection failed",
        "File not found",
        "Authentication failed",
        "Database connection failed",
        "Server timeout"
    ]
}

start_time = datetime(2026, 8, 20, 8, 0, 0)

with open("app.log", "w") as f:
    for line in range(100):

        timestamp = start_time + timedelta(
            minutes=random.randint(0, 12 * 60)
        )

        level = random.choice(levels)

        message = random.choice(messages[level])

        f.write(
            f"{timestamp.strftime('%Y-%m-%d %H:%M:%S')} "
            f"{level} {message}\n"
        )

print("Log file created successfully.")
```

```python
#read the log file and perform analysis on it
import re
from datetime import datetime
from collections import Counter, defaultdict

pattern = r"^(\d{4}-\d{2}-\d{2} \d{2}:\d{2}:\d{2}) (INFO|WARN|ERROR) (.*)$"

level_counts = Counter()
hour_counts = Counter()
error_messages = Counter()

logs_by_level = defaultdict(list)
error_times = []

with open("app.log", "r") as f:
    for line in f:
        line = line.strip()

        match = re.match(pattern, line)

        if match:
            timestamp_string = match.group(1)
            level = match.group(2)
            message = match.group(3)

            timestamp = datetime.strptime(
                timestamp_string,
                "%Y-%m-%d %H:%M:%S"
            )

            level_counts[level] += 1
            hour_counts[timestamp.hour] += 1
            logs_by_level[level].append(message)

            if level == "ERROR":
                error_messages[message] += 1
                error_times.append(timestamp)

print("Entries per level:")
print(level_counts)

print("\nBusiest hour:")
print(hour_counts.most_common(1))

print("\nMost common ERROR:")
print(error_messages.most_common(1))

print("\nGrouped logs:")
for level, messages_list in logs_by_level.items():
    print(level, ":", len(messages_list))

error_times.sort()

longest_gap = None

for i in range(1, len(error_times)):
    gap = error_times[i] - error_times[i - 1]

    if longest_gap is None or gap > longest_gap:
        longest_gap = gap

print("\nLongest ERROR gap:")
print(longest_gap)
```
