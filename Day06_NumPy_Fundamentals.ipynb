# Task-3

```python
import numpy as np
import matplotlib.pyplot as plt
import timeit

sizes = [1000, 10000, 100000, 1000000]

square_speedup = []
sum_speedup = []
dot_speedup = []


for n in sizes:

    lst = list(range(n))
    arr = np.array(lst)

    def python_square():
        result = []
        for x in lst:
            result.append(x * x)
        return result

    def numpy_square():
        return arr * arr

    python_time = timeit.timeit(python_square, number=10)
    numpy_time = timeit.timeit(numpy_square, number=10)

    square_speedup.append(python_time / numpy_time)


    def python_sum():
        total = 0
        for x in lst:
            total = total + x
        return total

    def numpy_sum():
        return arr.sum()

    python_time = timeit.timeit(python_sum, number=10)
    numpy_time = timeit.timeit(numpy_sum, number=10)

    sum_speedup.append(python_time / numpy_time)


    def python_dot():
        total = 0
        for i in range(n):
            total = total + lst[i] * lst[i]
        return total

    def numpy_dot():
        return np.dot(arr, arr)

    python_time = timeit.timeit(python_dot, number=10)
    numpy_time = timeit.timeit(numpy_dot, number=10)

    dot_speedup.append(python_time / numpy_time)


print("Square:", square_speedup)
print("Sum:", sum_speedup)
print("Dot Product:", dot_speedup)



plt.plot(sizes, square_speedup, label="Square")
plt.plot(sizes, sum_speedup, label="Sum")
plt.plot(sizes, dot_speedup, label="Dot Product")

plt.xlabel("Array Size")
plt.ylabel("Speedup")
plt.title("Python vs NumPy")

plt.legend()
plt.show()
```

# Task-6

```python
rng = np.random.default_rng(42)
marks = rng.integers(30, 101, size=(100, 5))
print("Marks shape:", marks.shape)
```

```python
student_average = marks.mean(axis=1)
print(student_average)
```

```python
exam_average = marks.mean(axis=0)

print(exam_average)
```

```python
passed = student_average >= 60
number_passed = np.sum(passed)
fraction_passed = number_passed / 100
print("Fraction passed:", fraction_passed)
```

```python
top10 = np.argsort(student_average)[-10:][::-1]
print(top10)
print(student_average[top10])
```

```python
exam_mean = marks.mean(axis=0)
exam_std = marks.std(axis=0)
standardised = (marks - exam_mean) / exam_std
print(standardised)
```

```python
new_marks = np.minimum(marks + 5, 100)
cap_count = np.sum(new_marks == 100)
print(new_marks)
print("Number of marks that reached 100:", cap_count)
```

```python
failed = marks < 60
failed_every_exam = np.all(failed, axis=1)
student_indices = np.where(failed_every_exam)[0]
print(student_indices)
```
