# Task-1

```python
class student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
        self.marks=[]

    def add_mark(self,score):
        self.marks.append(score)

    def average(self):
        return sum(self.marks)/len(self.marks)

    def highest(self):
        return max(self.marks)

    def lowest(self):
        return min(self.marks)

student1=student('abc',20)
student2=student('xyz',23)
student3=student('pqr',19)


student1.add_mark(80)
student1.add_mark(60)
student1.add_mark(70)
student1.add_mark(89)

student2.add_mark(90)
student2.add_mark(69)
student2.add_mark(76)
student2.add_mark(88)

student3.add_mark(99)
student3.add_mark(86)
student3.add_mark(56)
student3.add_mark(77)

print(student1.name,student1.age,student1.marks,student1.highest(),student1.average(),student1.lowest())
print(student2.name,student2.age,student2.marks,student2.highest(),student2.average(),student2.lowest())
print(student3.name,student3.age,student3.marks,student3.highest(),student3.average(),student3.lowest())
    
```

```python
#__str__ : means how we need organize the data . we give a syntex
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll
        self.marks = []
    def __str__(self):
        return f"name: {self.name} roll: {self.roll}"
s1 = Student("Asha", 12)

print(s1)
```

```python
#__repr__:when we pass a list of student then we use 
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll
        self.marks = []

    def __str__(self):
        return f"name: {self.name}, roll: {self.roll}"

    def __repr__(self):
        return f"name:{self.name},roll:{self.roll}"

    def __len__(self):
        return len(self.marks)

    def __eq__(self):
        return self.roll==self.roll

s1 = Student("Asha", 12)
s2 = Student("Nidhi", 13)

print(s1)
print(s2)

students = [s1, s2]
print(students)
```
