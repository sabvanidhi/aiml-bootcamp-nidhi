```python
import csv
import random
from datetime import datetime, timedelta

random.seed(42)

names=["Abc", "Def", "Xyz", "Pqr", "Aaa"]
domains=["gmail.com", "yahoo.com", "outlook.com"]

data=[]

data.append([
    "roll", "name", "email", "enrolled",
    "exam1", "exam2", "exam3", "exam4"
])

for roll in range(1, 201):

    name=random.choice(names)

    email=name.lower() + str(roll) + "@" + random.choice(domains)

    date=datetime(2026, 1, 1) + timedelta(
        days=random.randint(0, 180)
    )

    enrolled=date.strftime("%Y-%m-%d")

    exam1=random.randint(30, 100)
    exam2=random.randint(30, 100)
    exam3=random.randint(30, 100)
    exam4=random.randint(30, 100)

    if roll==10:
        exam2 = ""

    if roll==50:
        exam3 = ""

    if roll==20:
        email = "wrongemail"

    if roll==30:
        name = name.upper()

    data.append([
        roll, name, email, enrolled,
        exam1, exam2, exam3, exam4
    ])

with open("records.csv", "w", newline="") as f:
    csv.writer(f).writerows(data)

print("records.csv created")
```

```python
import csv
from collections import Counter, defaultdict
from datetime import datetime


def load_records(path):

    with open(path, "r") as f:
        data = list(csv.DictReader(f))

    print("total students:", len(data))

    domains = []
    wrong_emails = []

    for row in data:

        email = row["email"]

        if "@" in email:
            domain = email.split("@")[1]
            domains.append(domain)
        else:
            wrong_emails.append(row["roll"])

    print("domain count:", Counter(domains))
    print("Wrong email rolls:", wrong_emails)

    for row in data:
        row["name"] = row["name"].strip().title()
    for row in data:
        row["enrolled"] = datetime.strptime(row["enrolled"],"%Y-%m-%d")
    months = []

    for row in data:
        month = row["enrolled"].strftime("%Y-%m")
        months.append(month)

    print("Monthly enrolments:", Counter(months))

    students = defaultdict(list)

    for row in data:
        email = row["email"]
        if "@" in email:
            domain = email.split("@")[1]
            students[domain].append(row)

    print("Students by domain:")

    for domain in students:
        print(domain, len(students[domain]))

    for row in data:
        if row["exam1"] == "":
            row["exam1"] = None

        if row["exam2"] == "":
            row["exam2"] = None

        if row["exam3"] == "":
            row["exam3"] = None

        if row["exam4"] == "":
            row["exam4"] = None

    return data
records = load_records("records.csv")
```

```python
marks = np.array([
    [
        0 if row["exam1"] is None else int(row["exam1"]),
        0 if row["exam2"] is None else int(row["exam2"]),
        0 if row["exam3"] is None else int(row["exam3"]),
        0 if row["exam4"] is None else int(row["exam4"])
    ]
    for row in records
])
print(marks)
print("Shape:", marks.shape)

student_avg = marks.mean(axis=1)
print("Student average:")
print(student_avg)

exam_avg = marks.mean(axis=0)
print("Exam average:")
print(exam_avg)


passed = student_avg >= 60
print("Passed:", passed.sum())
print("Pass percentage:", passed.mean() * 100)


top10 = np.argsort(student_avg)[-10:][::-1]
print("Top 10:")
print(np.array([records[i]["name"] for i in top10]))
print("Hardest exam:", np.argmin(exam_avg) + 1)

spread = marks.std(axis=0)
print("Widest spread exam:", np.argmax(spread) + 1)

standard = (marks - marks.mean(axis=0)) / marks.std(axis=0)
print("Standardised marks:")
print(standard)

curve = np.minimum(marks + 5, 100)
print("Curved marks:")
print(curve)
print("Marks reached 100:", (curve == 100).sum())
```
