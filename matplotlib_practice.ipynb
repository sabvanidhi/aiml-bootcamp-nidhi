```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
fig, ax=plt.subplots(figsize=(5, 4))
x=np.linspace(0, 4*np.pi, 200)
ax.plot(x, np.sin(x), label='sin')
ax.plot(x, np.cos(x), label='cos')
ax.set_title('Sine and Cosine Waves')
ax.legend()

```

```python
np.random.seed(42)
x = np.random.rand(100) * 10
y = 2 * x + np.random.randn(100) * 2
fig, ax=plt.subplots()
ax.scatter(x, y)
ax.set_title("X and Y Relationship")
ax.set_xlabel("X")
ax.set_ylabel("Y")
plt.show()
```

```python
np.random.seed(42)
x=np.random.rand(100) * 10
noise=np.random.randn(100) * 2
y=2 * x + noise
third=np.random.rand(100) * 100
fourth=np.random.rand(100) * 180 + 20
fig, ax=plt.subplots(figsize=(5, 4))
scatter=ax.scatter(x,y,c=third,s=fourth,alpha=0.7)
ax.set_title("Four-Variable Scatter Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")
fig.colorbar(scatter, ax=ax, label="Third Variable")
plt.show()
```

```python
np.random.seed(42)
x = np.linspace(0, 10, 100)
y = 2 * x + np.random.normal(0, 2, 100)
categories = ["A", "B", "C", "D"]
values = [25, 40, 30, 50]
hist_data = np.random.normal(50, 10, 1000)

fig, axes = plt.subplots(2, 2, figsize=(11, 8))
axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title("Line Plot")
axes[0, 0].set_xlabel("X (units)")
axes[0, 0].set_ylabel("Value (units)")
axes[0, 1].scatter(x, y)
axes[0, 1].set_title("Scatter Plot")
axes[0, 1].set_xlabel("X (units)")
axes[0, 1].set_ylabel("Y (units)")
axes[1, 0].bar(categories, values)
axes[1, 0].set_title("Bar Chart")
axes[1, 0].set_xlabel("Category")
axes[1, 0].set_ylabel("Value (units)")
axes[1, 1].hist(hist_data, bins=30, edgecolor="white")
axes[1, 1].set_title("Histogram")
axes[1, 1].set_xlabel("Value (units)")
axes[1, 1].set_ylabel("Frequency (count)")

fig.suptitle("Four Matplotlib Plot")

plt.tight_layout()
plt.show()
```

# Task-6

```python
np.random.seed(42)

n = 200

study_hours = np.random.uniform(1, 10, n)
attendance = np.random.uniform(60, 100, n)

exam_score = (20+ 5 * study_hours+ 0.3 * attendance+ np.random.normal(0, 5, n))
passed = exam_score >= 50
failed = np.logical_not(passed)
pass_count = sum(passed)
fail_count = sum(failed)
pass_average = exam_score[passed].mean()
fail_average = exam_score[failed].mean()
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].scatter(study_hours[passed],exam_score[passed],color="green",label="Passed")
axes[0, 0].scatter(study_hours[failed],exam_score[failed],color="red",label="Failed")
axes[0, 0].set_title("Study Hours vs Exam Score")
axes[0, 0].set_xlabel("Study Hours")
axes[0, 0].set_ylabel("Exam Score")
axes[0, 0].legend()
axes[0, 1].hist(exam_score[passed],color="green",label="Passed")
axes[0, 1].hist(exam_score[failed],color="red",label="Failed")
axes[0, 1].set_title("Exam Score Distribution")
axes[0, 1].set_xlabel("Exam Score")
axes[0, 1].set_ylabel("Number of Students")
axes[0, 1].legend()
axes[1, 0].bar(["Passed", "Failed"],[pass_count, fail_count],color=["green", "red"])
axes[1, 0].set_title("Pass vs Fail")
axes[1, 0].set_xlabel("Result")
axes[1, 0].set_ylabel("Number of Students")
axes[1, 1].bar(["Passed", "Failed"],[pass_average, fail_average],color=["green", "red"]
)
axes[1, 1].set_title("Average Exam Score")
axes[1, 1].set_xlabel("Result")
axes[1, 1].set_ylabel("Average Score")
plt.tight_layout()
plt.savefig("report_figure.png",dpi=150,bbox_inches="tight")
plt.show()
```
