```python
abc=[10,20,30,40,50]
#loop
for n in abc:
    a=n*n
    print(a)
#functional
a=[n*n for n in abc]
print(a)
```

```python
a=['python ','html','css','java']
b=[]
c=[]
#loop
for n in a:
    b.append(n.upper())
print(b)

for n in a:
    c.append(n.strip())
print(c)

# functional

abc=list(n.upper() for n in a)

pqr=[n.strip() for n in a]
print(abc)
print(pqr)
```

```python
a=[1,2,3,4]
b=[5,6,7,8]
#loop


#functional
z=list(map(lambda a,b:a*b,a,b))

print(z)
```

```python
b=['python','css','html','java','mathametic']
#loop
#functional
a=list(filter(lambda word: len(word) >  4,b))
print(a)
```

```python
from functools import reduce
#loop
#functional
a=[10,20,30,40,50,60]

b=reduce(lambda x,y:x+y,a)
print(b)
```

```python
list=[
    {"item":"abc","price":30},
    {"item":"xyz","price":90},
    {"item":"aaa","price":76},
    {"item":"bbb","price":40}
    ]
#loop
#functional
b=sum(map(lambda list:list["price"],list))

print(b)
```

```python
from itertools import chain
#loop
#functional
print(list(chain([1, 2], [3, 4], [5])))
```

```python
from collections import Counter
#loop
#functional
b=['python','css','html','java','mathametic']

a=dict(map(lambda x:(x,len(x)),b))

print(a)
```

```python
l1=["",None,[],'abc','def','xyz']
#loop
#functional
list(filter(None,l1))
```

```python
from itertools import accumulate
abc=[10,20,30,40,50,60]
#loop
#functional
print(list(accumulate(abc)))
```
