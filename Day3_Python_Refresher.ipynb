```python
from collections import Counter
import string

with open("abc.txt", "r") as f:
    text = f.read().lower()

    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )
    words = text.split()
    stopwords = {
    "the", "a", "an", "is", "and",
    "of", "to", "in", "for", "on"
}
new_words = []
for i in words:
    if i not in stopwords:
        new_words.append(i)

words = new_words
counter = Counter(words)
print(counter)
print(counter.most_common(3))
```

```python
from itertools import combinations,permutations,product
student=["abc","xyz","pqr","lmn","def"]
print("Combinations of 2 students:")
print(list(combinations(student, 2)))
print("Combinations of 3 students:")
print(list(combinations(student, 3)))
print("Permutations of 2 students:")
print(list(permutations(student, 2)))
print("Product of colors and sizes:")
print(list(product([["red"], ["blue"], ["green"]],[["small"], ["medium"], ["large"]])))
```

```python
import re

text = """
Casha.sharma@codetrade.io wrote to ravi_k99@gmail.com
meera.p+work@company.co.in, phone 022-2555-1234
dev@sub.domain.example.org and +91 98765 43210
not.an.email@ nor @nothing.com -- watch these
"""

emails = re.findall(r"[\w.+-]+@[\w-]+\.[\w.]+", text)
print("emails found:")
for e in emails:
    print("  ", e)


phones = re.findall(
    r'(?:\+91[\s-]?\d{5}[\s-]?\d{5}|\b0\d{2,4}[-\s]\d{3,4}[-\s]\d{4}\b)',
    text
)
print(phones)


parts = re.findall(r'([\w.+-]+)@([\w.-]+\.\w+)', text)
print("\nUsername and domain:")
print(parts)

domains = [domain for username, domain in parts]
domain_counts = Counter(domains)
print("\nDomain counts:")
print(domain_counts)
```

```python
from collections import defaultdict

employees = [
    ("IT", "abc"),
    ("HR", "def"),
    ("IT", "xyz"),
    ("Sales", "pqr"),
    ("HR", "aaa")
]

departments = defaultdict(list)

for department, employee in employees:
    departments[department].append(employee)

print(departments)
```

```python
from collections import namedtuple

Point = namedtuple("Point", ["name", "age", "city"])

people = [
    Point("abc", 20, "Ahmedabad"),
    Point("xyz", 22, "Delhi"),
    Point("pqr", 29, "Mumbai"),
    Point("aba", 30, "Jaipur")
]

for p in people:
    print(p)
    print("p.name =", p.name, "p.age =", p.age, "p.city =", p.city)
    print("still a tuple:", p[0], p[1], p[2])
    print("unpacks like one:", *p)
    print()  
```
