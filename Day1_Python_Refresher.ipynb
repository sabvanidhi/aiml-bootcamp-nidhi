```python
def reverse_words(sentence):
    words = sentence.split()
    words = words[::-1]
    return ' '.join(words)

sentence = input("Enter a sentence: ")
print(reverse_words(sentence))
```

```python
def count_word(sentence):
    words=sentence.split()
    count= {}

    for i in words:
        if i in count:
            count[i]+=1
        else:
            count[i]=1
    return count

sentence = input("Enter a sentence: ")
print(count_word(sentence))
```
