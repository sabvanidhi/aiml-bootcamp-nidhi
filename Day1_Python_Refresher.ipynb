```python
for i in range(1, 101):
    if i%15==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")
    else:
        print(i)
```

```python
for i in range(1, 101):
    if i%15==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")
    elif i%7==0:
        print("FizzBuzzBazz")
    else:
        print(i)
```

```python
result = [
    "FizzBuzz" if i%15==0
    else "Fizz" if i%3==0
    else "Buzz" if i%5==0
    else i
    for i in range(1,101)
]

print(result)
```
